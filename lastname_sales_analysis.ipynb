What is monthly total revenue for in-store sales in each of the two sales territories, over the full
period covered by the data?


```python
current_sales = sales_df.copy()
# Defining the current_sales

current_sales['Transaction Date'] = pd.to_datetime(current_sales['Transaction Date'])
current_sales['Month'] = current_sales['Transaction Date'].dt.to_period('M')
current_sales['Year'] = current_sales['Transaction Date'].dt.year
# Convert the date and create the 'Month' column

monthly_rev = (current_sales
               .groupby('Month')['Sale Amount']
               .sum()
               .reset_index())

monthly_rev = monthly_rev.rename (columns = {'Sale Amount': 'Monthly Revenue'})
# Group by the month and sum of the sales amount

display(monthly_rev)
# Display the output


```

```python
# Merge sales with store details
richards_stores = sales_df.merge(
    stores_df[['Store ID', 'Territory Manager']],
    on='Store ID',
    how='left')

# Filter manager name (precaution)
richards_stores = richards_stores[
    richards_stores['Territory Manager']
    .str.strip()
    .str.contains('Richards', case=False, na=False)]

# Converting dates
richards_stores['Transaction Date'] = pd.to_datetime(
    richards_stores['Transaction Date'])

# Month column
richards_stores['Month'] = (
    richards_stores['Transaction Date'].dt.to_period('M'))

# Monthly revenue
monthly_rev = (
    richards_stores
    .groupby('Month')['Sale Amount']
    .sum()
    .reset_index())

monthly_rev = monthly_rev.rename(
    columns={'Sale Amount': 'Monthly Revenue'})

#Display the output
display(monthly_rev)
```

How would you rank the sales performance of each store in each sales territory? Which are the
top-performing stores?

```python
# Merging of necessary sales information (sales_df & stores_df)
df = sales_df.merge(
    stores_df[['Store ID', 'Territory Manager', 'Store Location']],
    on='Store ID',
    how='left')

# Total rounded sales per store
store_perf = (
    df.groupby(['Territory Manager', 'Store ID', 'Store Location'])['Sale Amount']
    .sum()
    .reset_index()
    .rename(columns={'Sale Amount': 'Total Sales'}))

store_perf['Total Sales'] = store_perf['Total Sales'].round(2)

# Rank stores within each territory
store_perf['Rank'] = store_perf.groupby('Territory Manager')['Total Sales'] \
                               .rank(ascending=False, method='first')

# Sort
store_perf = store_perf.sort_values(['Rank', 'Territory Manager'])

# Display Output
display(store_perf)
```

```python
customers_df = pd.read_csv('customer_list.csv', sep='|')

# Extracting 'First Name' and 'Last Name' from the 'name' column
# Being that 'name' column contains 'First Name Last Name' format
customers_df[['First Name', 'Last Name']] = customers_df['name'].str.split(' ', n=1, expand=True)

# Merging sales information with store territory info
sales_territory = sales_df.merge(
    stores_df[['Store ID', 'Territory Manager']],
    on='Store ID',
    how='left')

# Merging to customers
full_data = sales_territory.merge(
    customers_df,
    left_on='RewardsID',
    right_on='cust_id',
    how='left')

# Total spending per customer per manager
customer_totals = (
    full_data.groupby(['Territory Manager', 'cust_id', 'First Name', 'Last Name'])['Sale Amount']
    .sum()
    .reset_index()
    .rename(columns={'Sale Amount': 'Total Sales'}))

# Ranking customers within every territory manager
customer_totals['Rank'] = customer_totals.groupby('Territory Manager')['Total Sales'] \
                               .rank(ascending=False, method='dense')

# Top customers
top_customers = customer_totals[customer_totals['Rank'] == 1]

# Display Output
display(top_customers)
```

What is the number of transactions per month by product category in each assigned territory?
What is total sales revenue per month by category? What might this tell you about the most
popular products, and where could there be opportunity for growth?

```python
# Build full dataset (making my joins)
df = sales_df.merge(
    stores_df[['Store ID', 'Territory Manager']],
    on='Store ID',
    how='left')

# Merging with product info
df = df.merge(
    products_df[['Prod Num', 'CategoryID']],
    on ='Prod Num',
    how='left')

# Merging with product category info
df = df.merge(
    products_categories_df[['CategoryID', 'Category']],
    on='CategoryID',
    how='left')

# Rename 'Category' to 'Category Name' to match the original intent
df = df.rename(columns={'Category': 'Category Name'})

# Month Column
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
df['Month'] = df['Transaction Date'].dt.to_period('M')

# Total number of transactions per month by category/territory
transactions_per_month = (
    df.groupby(['Month', 'Territory Manager', 'Category Name'])['Sale Amount']
    .count()
    .reset_index()
    .rename(columns={'Sale Amount': 'Number of Transactions'}))

# Monthly Revenue

monthly_rev = (
    df.groupby(['Month', 'Territory Manager', 'Category Name'])['Sale Amount']
    .sum()
    .reset_index()
    .rename(columns={'Sale Amount': 'Monthly Revenue'}))

# Display Output
display(transactions_per_month, monthly_rev)
```

```python
import matplotlib.pyplot as plt

# Aggregation
transactions = (
    df.groupby(['Month', 'Category Name'])
    .size()
    .reset_index(name='Transaction Count'))

# Plotting
pivot_tx = transactions.pivot(index='Month', columns='Category Name', values='Transaction Count')

pivot_tx.plot(kind='line', marker='o')

plt.title('Monthly Transaction Count by Category')
plt.xlabel('Month')
plt.ylabel('Transaction Count')
plt.legend(title='Category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
revenue = (
    df.groupby(['Month', 'Category Name'])['Sale Amount']
    .sum()
    .reset_index(name='Total Revenue'))

pivot_rev = revenue.pivot(index='Month', columns='Category Name', values='Total Revenue')

pivot_rev.plot(kind='line', marker='o')

plt.title('Monthly Revenue by Category')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.legend(title='Category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
